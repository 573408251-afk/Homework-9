Задание 1
Датасет для домашнего задания вы найдете в разделе “Материалы к домашнему заданию” к этому занятию.
Определите, какому фильму было выставлено больше всего оценок 5.0.

```python
import pandas as pd
movies = pd.read_csv('movies.csv', encoding = 'utf-8')
ratings = pd.read_csv('ratings.csv', encoding = 'utf-8')

# Объединим таблицы по movieId
merged_data = pd.merge(movies, ratings, on='movieId', how = 'left')

# В объединенной таблицы оставляем фильмы с рейтингом 5
max_ratings = merged_data[merged_data['rating'] == 5.0]

# Группируем фильм и считаем количество оценок
top_movies = max_ratings.groupby('title').size().reset_index(name='count')

# Сортируем по убыванию количество фильмов с оценкой 5
top_movies = top_movies.sort_values(by='count', ascending = False)

# Выводим фильм с максимальной оценкой
print(top_movies.head(1))
```

```python
import pandas as pd
power = pd.read_csv('power.csv', encoding = 'utf-8')

# Фильтр по годам, категориям, странам Прибалтики
filtered_data = power[(power['year'] >= 2005) & (power['year'] <= 2010) & (power['category'].isin([4, 12, 21])) & (power['country'].isin(['Latvia', 'Lithuania', 'Estonia'])) & (power['quantity'] >= 0)]

# Суммарное потребление стран Прибалтики
consamption_country = filtered_data['quantity'].sum()
print(f'Суммарное потребление стран Прибалтики за период с 2005 по 2010 год: {int(consamption_country)}') 
```
